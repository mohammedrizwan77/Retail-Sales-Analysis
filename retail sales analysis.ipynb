```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
```

```python
df["order_date"]=pd.to_datetime(df["order_date"])
df["ship_date"]=pd.to_datetime(df["ship_date"])
```

```python
# 1. Fill from the same customer's other orders
# because repeated order id is low and repeated customers is high

df["segment"] = df["segment"].fillna(df.groupby("customer_id")["segment"].transform("first"))
```

```python
df.isna().sum()
```

```python
df["payment_method"]=df["payment_method"].fillna(df.groupby("customer_id")["payment_method"].transform("first"))
```

```python
df.isna().sum()
```

```python
# All rows whose order_id appears more than once (both copies shown)
dups = df[df["order_id"].duplicated(keep=False)].sort_values("order_id")
dups
```

```python
df["customer_rating"]=df["customer_rating"].fillna(df.groupby("order_id")["customer_rating"].transform("mean"))
df["customer_rating"]=df["customer_rating"].fillna(df.groupby("customer_id")["customer_rating"].transform("mean"))
```

```python
df.isna().sum()
```

```python
df["customer_rating"]=df["customer_rating"].replace(["N/A","-","null"],np.nan)
df["customer_rating"]=pd.to_numeric(df["customer_rating"])
df["customer_rating"]=df["customer_rating"].fillna(df["customer_rating"].mean().round(0))
```

```python
df["customer_rating"]=df["customer_rating"].round(1)
```

```python
df.isna().sum()
```

```python
df=df.dropna(subset=["segment","payment_method"])
```

```python
df.to_csv("retail_sales_data.csv",index=False)
```

```python
# Convert numeric columns in case they were read as text
numeric_cols = [
    "unit_price", "quantity", "discount", "sales",
    "profit", "customer_rating"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Add useful time fields
df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.to_period("M").dt.to_timestamp()
df["weekday"] = df["order_date"].dt.day_name()
df["shipping_days"] = (df["ship_date"] - df["order_date"]).dt.days

sns.set_theme(style="whitegrid")

# Basic overview
print("Rows and columns:", df.shape)
print("\nMissing values:")
print(df.isna().sum().sort_values(ascending=False).head(2))
print("\nOverall totals:")
print(df[["sales", "profit", "quantity"]].sum())
print("\nAverage customer rating:", round(df["customer_rating"].mean(), 2))
```

```python

# Year-over-year totals
print("\nSales and profit by year:")
print(df.groupby("year")[["sales", "profit"]].sum().round(2))

# Compare business dimensions
dimensions = [
    "segment", "region", "category", "product",
    "ship_mode", "payment_method"]

for col in dimensions:
    summary = (
        df.groupby(col, dropna=False)
          .agg(orders=("order_id", "nunique"),
               sales=("sales", "sum"),
               profit=("profit", "sum"),
               avg_rating=("customer_rating", "mean"))
          .sort_values("sales", ascending=False)
          .round(2))
    print(f"\nPerformance by {col}:")
    print(summary)
```

```python
# product categories generate the most sales and profit
top_category=df.groupby(["category"])[["sales","profit"]].sum().sort_values(by="sales",ascending=False)
# top_category
# plt.bar(top_category.index, top_category.values)
sns.set_theme(style="whitegrid")
fig,axes=plt.subplots(1,2,figsize=(8,3))
a=sns.barplot(data=top_category,x="category",y="sales",errorbar=None,ax=axes[0])
for v1 in a.containers:
    a.bar_label(v1)
axes[0].set_title("profit")
b=sns.barplot(data=top_category,x="category",y="profit",errorbar=None,ax=axes[1])
for v1 in b.containers:
    b.bar_label(v1)
axes[1].set_title("sales")
```

```python
# individual products have the highest and lowest sales and profit
product_summary=df.groupby("product")[["profit","sales"]].sum()
# print("Highest total sales:")
# print(product_summary["sales"].idxmax(), product_summary["sales"].max())

# print("\nLowest total sales:")
# print(product_summary["sales"].idxmin(), product_summary["sales"].min())

# print("\nhighest total profit")
# print(product_summary["profit"].idxmax(),product_summary["profit"].max())

# print("\nlowest total profit")
# print(product_summary["profit"].idxmin(),product_summary["profit"].min())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

product_summary["sales"].sort_values().plot(
    kind="barh", ax=axes[0], color="steelblue"
)
axes[0].set_title("Total Sales by Product")
axes[0].set_xlabel("Sales")
axes[0].set_ylabel("Product")

product_summary["profit"].sort_values().plot(
    kind="barh", ax=axes[1], color="seagreen"
)
axes[1].set_title("Total Profit by Product")
axes[1].set_xlabel("Profit")
axes[1].set_ylabel("")

plt.tight_layout()
plt.show()
```

```python
# sales and profit change over time by month, quarter, or year

monthly = (df.set_index("order_date")[["sales", "profit"]].resample("ME").sum())

quarterly = (df.set_index("order_date")[["sales", "profit"]].resample("QE").sum())

yearly = (df.set_index("order_date")[["sales", "profit"]].resample("YE").sum())

# print("Monthly totals:\n",monthly)
# print("Quarterly totals:\n", quarterly)
# print("Yearly totals:\n", yearly)
fig, axes = plt.subplots(3, 2, figsize=(10,8))

for row, (period, summary) in enumerate([
    ("Monthly", monthly),
    ("Quarterly", quarterly),
    ("Yearly", yearly),
]):
    summary["sales"].plot(ax=axes[row, 0], marker="o", color="steelblue")
    axes[row, 0].set_title(f"{period} Sales")
    axes[row, 0].set_ylabel("Sales")

    summary["profit"].plot(ax=axes[row, 1], marker="o", color="seagreen")
    axes[row, 1].set_title(f"{period} Profit")
    axes[row, 1].set_ylabel("Profit")

plt.tight_layout()
plt.show()
```
